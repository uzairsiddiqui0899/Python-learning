```python
subset_loc = df.loc[1:3, ['Name', 'Salary']] #all 1,2,3 are included
print("Subset with .loc:\n", subset_loc )
```

### Handling Missing Data

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={"Category": "category",
           "Region" : "category"}
    )
df.head()
```

```python
df.info()
```

```python
# Detect missing values
print(df.isna().sum())
```

```python
# Fill missing values with mean
df['Quantity'] = pd.to_numeric(df['Quantity'])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
```

```python
# Fill missing vales with median
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
```

```python
# Fill missing values with mode
df['Category'] = df['Category'].fillna(df['Category'].mode()[0])
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.dropna(subset=['Region'],inplace=True)
df.isna().sum()
```

```python
df.info()
```

### Sorting and Ranking

```python
import pandas as pd

df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={"Category": "category",
           "Region" : "category"}
    )
df.head()
```

```python
#Sort by Quantity descending and Sales ascending

df_sorted = df.sort_values(by=['Quantity', 'Sales'], ascending=[False, True])
print("Sorted the DataFrame")
df_sorted.head()
```

```python
# apply() - Create revenue category
def revenue_category(row):
  if row["revenue"] > 500:
    return "High"
  elif row["revenue"] >= 200:
    return "Medium"
  else:
    return "Low"

df["revenue_category"] = df.apply(revenue_category, axis=1)
```

```python
# Verify
df["revenue_category"].value_counts()
```

```python
#Validate logic manually

df[["revenue", "revenue_category"]].sample(10)

# Revenue > 500 -- High
# Between 200-500 -- Medium
# < 200 -- Low
```

```python
#pipe() - Method chaining
def add_profit(data):
  data["profit"] = data["revenue"] * 0.2
  return data

def add_net_revenue(data):
  data["net_revenue"] = data["revenue"] - data["profit"]
  return data

df_pipe = df.pipe(add_profit).pipe(add_net_revenue)
```

### Outlier Removal

```python
#IQR method

Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = df[(df["revenue"] < lower_bound) | (df["revenue"] > upper_bound)]

outliers_iqr.shape #Verification
```

```python
outliers_iqr[["revenue", "product_category"]].head() #inspect
```

```python
df = df[(df["revenue"] >= lower_bound) & (df["revenue"] <= upper_bound)]
df.shape
```

```python
# Z-Score method

from scipy.stats import zscore

df = pd.read_csv("ecommerce_sales_analytics_5000.csv")
df["revenue_zscore"] = zscore(df["revenue"])
outliers_zscore = df[df["revenue_zscore"].abs() > 3]

# Z-score shows, how many standard deviations a value is from the mean
# Rule: |Z| > 3 -- outlier

outliers_zscore.shape
```

```python
df = df[df["revenue_zscore"].abs() <= 3] #removing outliers
df.shape
```
