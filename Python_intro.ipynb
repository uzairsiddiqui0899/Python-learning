```python
class Student:
  def __init__(self, name, age, grade):
    self.name = name
    self.age = age
    self.grade = grade

  def display_info(self):
    print(f"Name: {self.name}")
    print(f"Age: {self.age}")
    print(f"Grade: {self.grade}")

  def is_eligible(self):
    if self.age >= 15:
      print(self.name, "is eligible for admission.")
    else:
      print(self.name, "is not eligible for admission.")

# Create instances/objects of the Student class
student1 = Student("Ali", 16, "10th")
student2 = Student("Sara", 14, "8th")

# Accessinf attributes
print(student1.name)
print(student2.grade)
```

```python
# calling methods
student1.display_info()
student1.is_eligible()

student2.display_info()
student2.is_eligible()
```

```python
# Using getters and setters

class Student:
  def grade(self):
    return self.__grade # double underscore is a way to encapsulate data
  def __init__(self, name, grade):
    self.__grade = grade
  def get_grade(self):
    return self.__grade
  def set_grade(self, grade):
    if 0 <= grade <= 100:
      self.__grade = grade
    else:
      print("Invalid grade. Grade should be between 0 and 100.")

student = Student("Ali", 90)
student.set_grade(95)
print(student.get_grade())
```

# Method Overriding

```python
class Animal:
  def speak(self):
    print("Some sound")

class Dog(Animal): #Its inheriting from class Animal but printing specifically Woof for Dog
  def speak(self):
    print("Woof! Woof!")

dog1 = Dog()
dog1.speak()
```

```python
# polymorphism
class Dog:
  def speak(self):
    print("Woof! Woof!")

class Cat:
  def speak(self):
    print("Meow! Meow!")

for pet in [Dog(), Cat()]:
  pet.speak()
```

```python
class Vehicle:
  def move(self):
    print("Vehicle is moving")

class Car(Vehicle):
  def move(self):
    print("Car is moving")

class Boat(Vehicle):
  def move(self):
    print("Boat is moving")

car1 = Car()
car1.move()

boat1 = Boat()
boat1.move()
```

# Python Special Methods

```python
class Book:
  def __init__(self, title, author):
    self.title = title
    self.author = author
  def __str__(self):
    return f"'{self.title}' by {self.author}"

book1 = Book("Python", "Uzair")
print(book1)
```

```python
class Playlist:
  def __init__(self, songs):
    self.songs = songs
  def __len__(self):
    return len(self.songs)
  def __add__(self, other):
    return Playlist(self.songs + other.songs)

playlist1 = Playlist(["Song1", "Song2"]) #2
print(len(playlist1))

playlist2 = Playlist(["Song3"])

print(len(playlist1 + playlist2)) #3
```

# Exercise 1

**Task 1:** Suppose a mobile phone shop has 5 sales employees. Write a Python program that uses a for loop to enter the daily
sales amount (in Rs.) generated by each employee. Calculate the tax using the following rules:                                                                  
• If the sales amount is less than Rs. 50,000, tax is 5%.  
• If the sales amount is between Rs. 50,000 and Rs. 100,000, tax is 10%.  
• If the sales amount is greater than Rs. 100,000, tax is 15%.  
For each employee, display:
• Sales Amount  
• Tax Amount  
After entering all 5 employees, display:
• Total Sales  
• Average Sales

```python
# Task 1:
total_sales = 0

for i in range(1, 6):
    sales = float(input(f"Enter daily sales for employee {i} (Rs.): "))

    # Decide tax rate based on the rules
    if sales < 50000:
        tax_rate = 0.05
    elif sales <= 100000:          # 50,000 to 100,000
        tax_rate = 0.10
    else:                          # greater than 100,000
        tax_rate = 0.15

    tax = sales * tax_rate

    print(f"Employee {i} -> Sales Amount: Rs. {sales:.2f}, Tax Amount: Rs. {tax:.2f}")
    print("-" * 45)

    total_sales += sales

average_sales = total_sales / 5

print(f"Total Sales:   Rs. {total_sales:.2f}")
print(f"Average Sales: Rs. {average_sales:.2f}")
```

**Task 3:** Create a Bank Account System using Object-Oriented Programming (OOP).
Create the following classes:
• Customer  
• Account  
The system should implement the following functionalities:
• Create a customer account.
• Deposit money.
• Withdraw money.
• Display the current balance.

```python
# Task 3:
class Customer:
    def __init__(self, name, customer_id):
        self.name = name
        self.customer_id = customer_id


class Account:
    def __init__(self, customer, balance=0):
        self.customer = customer
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited Rs. {amount}. New balance: Rs. {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.balance -= amount
            print(f"Withdrew Rs. {amount}. New balance: Rs. {self.balance}")

    def display_balance(self):
        print(f"Account holder: {self.customer.name} (ID: {self.customer.customer_id})")
        print(f"Current balance: Rs. {self.balance}")


# --- Using the system ---
c1 = Customer("Uzair", 1001)
acc1 = Account(c1, 10000)      # opening balance Rs. 1000

acc1.display_balance()
acc1.deposit(50000)
acc1.withdraw(20000)
acc1.withdraw(100000)          # this will show insufficient balance
acc1.display_balance()
```
